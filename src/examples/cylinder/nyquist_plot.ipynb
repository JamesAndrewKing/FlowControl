```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Load data
cwd =  Path.cwd()
path_in = cwd / "data_output" / "nyquist"
G_vals = np.load(path_in / "G_vals.npy")
omega_range = np.load(path_in / "omega_range.npy")

def plot_nyquist(K=1.0):
    KG_vals = K * G_vals
    plt.figure(figsize=(6,6))
    plt.plot(KG_vals.real, KG_vals.imag, label=f"Nyquist curve (K={K})")
    plt.plot(KG_vals.real, -KG_vals.imag, '--', color='gray', label="Negative freq (conj)")
    plt.scatter([-1], [0], color='red', label='-1 (critical point)')
    plt.xlabel('Re(KG(iω))')
    plt.ylabel('Im(KG(iω))')
    plt.title(f'Nyquist Plot (K={K})')
    plt.legend()
    plt.grid(True)
    plt.axis('equal')
    plt.show()

# Try different K interactively
plot_nyquist(K=1.0)
plot_nyquist(K=2.0)
plot_nyquist(K=0.5)
```

```python
import ipywidgets as widgets
from ipywidgets import interact

@interact(K=(0.1, 100.0, 0.1))
def interactive_nyquist(K=1.0):
    plot_nyquist(K)
```
